# **Project Summary -**

The project "Yes Bank Stock Price Prediction Using Machine Learning" focuses on analyzing historical stock market data and predicting the closing stock price using supervised machine learning regression algorithms. The main objective of this project is to understand historical stock price patterns, identify relationships between different stock market variables, and develop machine learning models that can estimate stock closing prices.

The project uses historical Yes Bank stock market data containing numerical features such as Open, High, Low, and Close prices. The Close price is considered the target variable for prediction. The project follows a structured machine learning workflow, starting with Exploratory Data Analysis (EDA), followed by data preprocessing, feature analysis, model implementation, evaluation, and hyperparameter tuning.

Initially, Exploratory Data Analysis was performed to understand the dataset's structure, statistical properties, and distribution of stock prices. Different visualizations, including line plots, correlation analysis, and boxplots, were used to identify patterns and relationships among the variables. Missing values and potential outliers were also examined. The boxplot analysis identified 9 potential outliers in the closing price data.

During data preprocessing, numerical features were prepared for machine learning. Since the dataset contained no categorical columns, categorical encoding was not required. Feature scaling was performed to bring numerical variables into a suitable range. The dataset was divided into training and testing sets using an 80:20 chronological splitting approach. A total of 147 observations were used for training and 37 observations for testing. Chronological splitting was selected because stock market data is time-dependent, and preserving the time order helps reduce the risk of future data leakage.

Three supervised machine learning regression approaches were implemented and evaluated: Linear Regression, Random Forest Regressor, and Gradient Boosting Regressor. Linear Regression was used as a baseline model to understand the relationship between input variables and the target variable. Random Forest Regressor combined multiple decision trees to generate predictions, while Gradient Boosting Regressor built an ensemble of trees sequentially to improve prediction performance.

The performance of the models was evaluated using four important regression metrics: Mean Absolute Error (MAE), Mean Squared Error (MSE), Root Mean Squared Error (RMSE), and R² Score. These metrics helped measure prediction errors and determine how effectively each model explained variations in stock prices.

Hyperparameter optimization was also performed using GridSearchCV to identify suitable model configurations. However, tuning did not improve every evaluation metric, demonstrating that hyperparameter optimization does not always guarantee better performance.

Among the implemented models, the original Gradient Boosting Regressor achieved the strongest overall performance, with an MAE of approximately 24.37, MSE of 1370.07, RMSE of 37.01, and R² Score of 0.9166. This indicates that the model explained approximately 91.66% of the variation in the test target values. Feature importance analysis showed that Low had the highest importance (approximately 0.329), followed by Close (0.207), High (0.167), and Open (0.009).

Finally, the best-performing Gradient Boosting model was saved using the Joblib library, and its model configuration and feature importance information were exported into a JSON file for reference.

Overall, this project provided practical experience in data analysis, data preprocessing, regression modeling, model evaluation, hyperparameter tuning, and model persistence. It demonstrates how machine learning techniques can be applied to historical financial data to estimate stock prices. However, the predictions are estimates based on historical patterns and should not be considered guaranteed future stock market outcomes.

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis (H₀):     There is no significant difference between the average opening price and average closing price of Yes Bank.

Alternative Hypothesis (H₁):  There is a significant difference between the average opening price and average closing price of Yes Bank.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
# Hypothesis Testing - 1

from scipy.stats import ttest_rel

# Perform Paired T-Test
t_stat, p_value = ttest_rel(df['Open'], df['Close'])

print("T-Statistic:", t_stat)
print("P-Value:", p_value)

# Check Hypothesis
if p_value < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
````

##### Which statistical test have you done to obtain P-Value?

I have performed a Paired T-Test to obtain the P-value.

##### Why did you choose the specific statistical test?

### Hypothetical Statement - 3

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

1. Research Hypothesis
Null Hypothesis (H₀): There is no statistically significant difference in the average closing stock prices of Yes Bank before 2018 and from 2018 onwards.

Alternative Hypothesis (H₁): There is a statistically significant difference in the average closing stock prices of Yes Bank before 2018 and from 2018 onwards.

Significance Level: 0.05 (5%)
2. Statistical Test
- Test Used: Independent Samples T-Test (Welch's T-Test)
- Group 1: Closing prices before 2018.
- Group 2: Closing prices from 2018 onwards.
- Purpose: To compare the average closing prices of the two independent time-period groups.
3. Interpretation
- If p-value is less than 0.05, reject the Null Hypothesis.
- If p-value is greater than or equal to 0.05, fail to reject the Null Hypothesis.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
# Hypothesis Testing - 3

from scipy.stats import ttest_ind

# Create Two Groups
group1 = df[df['Date'].dt.year < 2018]['Close']

group2 = df[df['Date'].dt.year >= 2018]['Close']

# Perform Independent T-Test
t_stat, p_value = ttest_ind(
    group1,
    group2,
    equal_var=False
)

print("Average Closing Price Before 2018:", group1.mean())
print("Average Closing Price From 2018:", group2.mean())

print("T-Statistic:", t_stat)
print("P-Value:", p_value)

# Check Hypothesis
if p_value < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
````

##### Which statistical test have you done to obtain P-Value?

I have performed an Independent T-Test (Welch's T-Test) to obtain the P-value.

##### Why did you choose the specific statistical test?

I selected the Independent T-Test because I wanted to compare the average closing stock prices of two independent groups: before 2018 and from 2018 onwards. Welch's T-Test is suitable because it does not require both groups to have equal variances.

````python
mae_gb_tuned = mean_absolute_error(y_test, y_pred_gb_tuned)
mse_gb_tuned = mean_squared_error(y_test, y_pred_gb_tuned)
rmse_gb_tuned = np.sqrt(mse_gb_tuned)
r2_gb_tuned = r2_score(y_test, y_pred_gb_tuned)

print("Tuned MAE:", mae_gb_tuned)
print("Tuned MSE:", mse_gb_tuned)
print("Tuned RMSE:", rmse_gb_tuned)
print("Tuned R2 Score:", r2_gb_tuned)
````

Original vs Tuned Model Comparison

````python
gb_comparison = pd.DataFrame({
    'Metric': ['MAE', 'MSE', 'RMSE', 'R2 Score'],
    'Original Model': [mae_gb, mse_gb, rmse_gb, r2_gb],
    'Tuned Model': [
        mae_gb_tuned,
        mse_gb_tuned,
        rmse_gb_tuned,
        r2_gb_tuned
    ]
})

print(gb_comparison)

gb_comparison.set_index('Metric').plot(
    kind='bar',
    figsize=(10, 5)
)

plt.title("Gradient Boosting: Original vs Tuned Model")
plt.ylabel("Metric Score")
plt.xticks(rotation=0)
plt.grid(axis='y')
plt.show()
````

### 1. Which Evaluation metrics did you consider for a positive business impact and why?

I considered MAE, MSE, RMSE, and R² Score to evaluate the performance of the machine learning models.

- MAE measures the average absolute prediction error and helps understand how far predictions are from actual stock prices.
- MSE penalizes larger prediction errors and helps identify significant prediction mistakes.
- RMSE represents prediction error in the original stock price units, making it easier to interpret.
- R² Score measures how much variation in stock prices is explained by the model.

These metrics help evaluate prediction accuracy, minimize financial estimation errors, and support better data-driven business decisions. I gave importance to lower MAE, MSE, RMSE and a higher R² Score.

### 3. Explain the model which you have used and the feature importance using any model explainability tool?

I selected Gradient Boosting Regressor as my final prediction model and used Permutation Feature Importance as a model explainability technique.

The feature importance analysis shows that Low has the highest importance score (0.3293), followed by Close (0.2072), High (0.1669), and Open (0.0095).

````python
# Feature Importance using Permutation Importance

from sklearn.inspection import permutation_importance
import pandas as pd
import matplotlib.pyplot as plt

# Calculate Feature Importance
result = permutation_importance(
    best_gb_model,
    X_test,
    y_test,
    n_repeats=20,
    random_state=42,
    scoring='r2'
)

# Create Feature Importance DataFrame
feature_importance = pd.DataFrame({
    'Feature': X_test.columns,
    'Importance': result.importances_mean
})

# Sort Features
feature_importance = feature_importance.sort_values(
    by='Importance',
    ascending=False
)

print("Feature Importance:")
print(feature_importance)

# Visualize Feature Importance
plt.figure(figsize=(9, 5))

plt.barh(
    feature_importance['Feature'],
    feature_importance['Importance']
)

plt.xlabel("Importance Score")
plt.ylabel("Features")
plt.title("Gradient Boosting - Feature Importance")
plt.gca().invert_yaxis()
plt.grid(axis='x')

plt.show()
````
